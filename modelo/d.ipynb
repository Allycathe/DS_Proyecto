### Librerias

```python
!pip install --upgrade pandas numpy scikit-learn matplotlib

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_squared_error, mean_absolute_percentage_error, r2_score
```

### Parametros

```python
DEBUG = True # Prints y funciones de prueba principalmente

# Ruta del dataset
dataset_src = "./" # CAMBIAR, integrar a la ruta el archivo .csv

# Model
model_test_split = 0.2 # Train: 0.8

# DBSCAN parametros / NO USADO
#dbscan_minPts = 1
#dbscan_epsilon = 1
```

### Cargar Dataset

```python
df = pd.read_csv(dataset_src)

if DEBUG:
    print(df.info())
    print(df.describe())
    print(df.isna().sum())
```

### Modelo

```python
y = df["ICVU"] # CORREGIR
X = df.drop(columns=["ICVU"])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=model_test_split, random_state=42
)
```

```python
# Escalar datos para KNN (necesario)
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsRegressor())
])
```

```python
# Buscar el mejor K / REVISAR
param_grid = {
    "knn__n_neighbors": range(1, 21),
    "knn__weights": ["uniform", "distance"],
    "knn__p": [1, 2]   # Manhattan vs Euclidiana
}
grid = GridSearchCV(pipe, param_grid, cv=5,
                    scoring="neg_root_mean_squared_error", n_jobs=-1)
grid.fit(X_train, y_train)

print("Mejores parámetros:", grid.best_params_)
```

### Evaluación del Modelo

```python
# REVISAR/CAMBIAR a otras metricas
y_pred = grid.predict(X_test)
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("MAE :", mean_absolute_error(y_test, y_pred))
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred))
print("R²  :", r2_score(y_test, y_pred))
```
